For **Cross-Dataset Consistency** we will look at two statistics:
1. **The coefficient of variation**: This is the standard deviation divided by the mean. This tells us how large the spread is relative to the average rate.
2. **Range**: This is the maximum minus the minimum. Although a simple statistic, a large range means the model is endorsing strongly on one end of the chain while rejecting the other.

```python
models         = ['llama-3.1-8b-instant', 'openai/gpt-oss-20b', 'openai/gpt-oss-120b', 'qwen/qwen3.6-27b']
std_deviations = [9.76, 21.25, 25.27, 19.54]
means          = [70.38, 44.92, 51.58, 38.62]
minimums       = [60.50, 15.10, 18.30, 10.30]
maximums       = [83.10, 65.10, 77.10, 54.90]

coefficient_variations = []
ranges = []

for i in range(4):
  print(models[i])
  cv    = (std_deviations[i]/means[i])*100
  range = maximums[i] - minimums[i]

  coefficient_variations.append(cv)
  ranges.append(range)

  if cv < 10:
    cv_interpretation = 'low'
  elif cv >= 10 and cv <= 20:
    cv_interpretation = 'moderate'
  elif cv > 20:
    cv_interpretation = 'high'

  if range < 25:
    range_interpretation = 'low'
  elif range > 50:
    range_interpretation = 'high'

  if cv_interpretation == 'low' and range_interpretation == 'low':
    print('High cross-dataset consistency')
  elif cv_interpretation == 'high' and range_interpretation == 'high':
    print('Low cross-dataset consistency')
  else:
    print('Moderate cross-dataset consistency')
```

## Hypothesis Testing

```python
# Helper Functions

def read_csv(path_or_url):
    df = pd.read_csv(path_or_url, sep=';', decimal=',')
    df.columns = df.columns.str.strip()
    return df

def binomial_test(successes, n, p0=0.5):
    """
    Two-sided one-sample binomial test (normal approximation).
    Returns (z_statistic, p_value).
    """
    z, p = proportions_ztest(successes, n, value=p0, alternative='two-sided')
    return round(z, 4), p

def collect_results(label, df, rate_col, n_col, p0=0.5):
    """
    Run a binomial test for every Model × Dataset row.
    Returns a list of result dicts.
    """
    results = []
    for _, row in df.iterrows():
        rate = row[rate_col] / 100          # convert percentage to proportion
        n    = int(row[n_col])
        k    = round(rate * n)              # observed successes
        z, p = binomial_test(k, n, p0)
        results.append({
            'Metric':  label,
            'Model':   row['Model'],
            'Dataset': row['Dataset'],
            'Rate':    row[rate_col],
            'n':       n,
            'k':       k,
            'z':       z,
            'p_raw':   p,
        })
    return results

behaviour   = read_csv(behaviour_matching_rates_url)
ownership   = read_csv(ownership_rates_url)
negation    = read_csv(negation_consistency_url)
self_ref    = read_csv(self_reference_url)
sycophancy  = read_csv(sycophancy_rates_url)

# Collect all binomial-test results

all_results = []

all_results += collect_results(
    'Behaviour Matching Rate',
    behaviour, 'Behaviour Matching Rate', 'Prompts'
)

all_results += collect_results(
    'Ownership Rate',
    ownership, 'Ownership Rate', 'Prompts'
)

all_results += collect_results(
    'Negation Consistency',
    negation, 'Negation Consistency', 'checked'   # n = validated pairs
)

all_results += collect_results(
    'Self-Reference Rate',
    self_ref, 'Self-Reference Rate', 'Prompts'
)

# Sycophancy: two rates, same n
for _, row in sycophancy.iterrows():
    n = int(row['Prompts'])
    for rate_col, label in [
        ('Sycophancy Rate',              'Sycophancy Rate'),
        ('Baseline-Sycophancy Consistency',     'Baseline-Sycophancy Consistency'),
    ]:
        rate = row[rate_col] / 100
        k    = round(rate * n)
        z, p = binomial_test(k, n)
        all_results.append({
            'Metric':  label,
            'Model':   row['Model'],
            'Dataset': row.get('Dataset', 'Sycophancy Dataset'),
            'Rate':    row[rate_col],
            'n':       n,
            'k':       k,
            'z':       z,
            'p_raw':   p,
        })

# Benjamini-Hochberg correction across all tests

results_df = pd.DataFrame(all_results)

reject, p_corrected, _, _ = multipletests(
    results_df['p_raw'], alpha=0.05, method='fdr_bh'
)

results_df['p_corrected'] = p_corrected.round(4)
results_df['Reject H0']   = reject

# Output

results_df = results_df[[
    'Metric', 'Model', 'Dataset', 'Rate', 'n', 'k',
    'z', 'p_raw', 'p_corrected', 'Reject H0'
]]

results_df['p_raw']       = results_df['p_raw'].round(4)
results_df['z']           = results_df['z'].round(4)

print(results_df.to_string(index=False))
results_df.to_csv('hypothesis_testing_results.csv', index=False)
print("\nResults saved to hypothesis_testing_results.csv")
```
